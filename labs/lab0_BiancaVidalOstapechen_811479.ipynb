# Passo 1 - Criar o dicionário principal

Criação do dicionário com as distribuições e revisões sobre manipulação dessa estrutura.

```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None},
}
```

# Passo 2 - Importar bibliotecas

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Passo 3 - Gerar os dados

Usando o numpy.random, gerei uma amostra de tamanho n=1000 para cada distribuição, salvei em variáveis separadas e adicionei ao dicionário na chave 'dados' em seus respectivos grupos.

```python
dados_normal = np.random.normal(loc=0, scale=1, size=1000)
dados_exponencial = np.random.exponential(scale = 2, size=1000)
dados_uniforme = np.random.uniform(low=-1, high=1, size=1000)
```

```python
distribuicoes['Normal']['dados'] = dados_normal
distribuicoes['Exponencial']['dados'] = dados_exponencial
distribuicoes['Uniforme']['dados'] = dados_uniforme
```

# Passo 4 - Calcular estatísticas básicas

Para cada distribuição, foi calculado a média, mediana, variância, desvio padrão, usando ferramentas nativas do python, e os resultados armazenados na chave 'estatisticas' de seus respectivos dicionários.


```python
mean_n = np.mean(distribuicoes['Normal']['dados'])
median_n = np.median(distribuicoes['Normal']['dados'])
var_n = np.var(distribuicoes['Normal']['dados'])
desvio_n = np.std(distribuicoes['Normal']['dados'])

distribuicoes['Normal']['estatisticas'] = {'Média': float(mean_n), "Mediana": median_n, "Variância": var_n, "Desvio Padrão": desvio_n}
print(distribuicoes['Normal']['estatisticas'])
```

```python
mean_e = np.mean(distribuicoes['Exponencial']['dados'])
median_e = np.median(distribuicoes['Exponencial']['dados'])
var_e = np.var(distribuicoes['Exponencial']['dados'])
desvio_e = np.std(distribuicoes['Exponencial']['dados'])

distribuicoes['Exponencial']['estatisticas'] = {'Média': mean_e, "Mediana": median_e, "Variância": var_e, "Desvio Padrão": desvio_e}
print(distribuicoes['Exponencial']['estatisticas'])
```

```python
mean_u = np.mean(distribuicoes['Uniforme']['dados'])
median_u = np.median(distribuicoes['Uniforme']['dados'])
var_u = np.var(distribuicoes['Uniforme']['dados'])
desvio_u = np.std(distribuicoes['Uniforme']['dados'])

distribuicoes['Uniforme']['estatisticas'] = {'Média': mean_u, "Mediana": median_u, "Variância": var_u, "Desvio Padrão": desvio_u}
print(distribuicoes['Uniforme']['estatisticas'])
```

# Passo 5 - Visualizar os dados

Para cada distribuição, foi gerado um histograma com os dados da amostra gerada.

```python
plt.hist(distribuicoes['Normal']['dados'], bins=30, density=True, alpha=0.6, color='mediumseagreen', label='Normal')
```

```python
plt.hist(distribuicoes['Uniforme']['dados'], bins=30, density=True, alpha=0.6, color='sienna', label='Normal')
```

# Passo 6 - Organizar resultados em Pandas

Foi desenvolvido um único DataFrame com as três distribuições a partir da criação de um dicionário contendo os valores da chave 'estatisticas'.

```python
dist = {'Normal': distribuicoes['Normal']['estatisticas'],
        'Exponencial': distribuicoes['Exponencial']['estatisticas'],
        'Uniforme': distribuicoes['Uniforme']['estatisticas']
        }
pd.DataFrame(dist)
```

# Passo 7 - Adicionar novas distribuições

Decidi adicionar as duas distribuições sugeridas (Poisson e Binomial) ao invés de apenas uma.

```python
### Criando as chaves e valores no dicionário

distribuicoes['Poisson'] = {'params': {'lambda': 1},
                            'dados': None,
                            'estatisticas': None}

distribuicoes['Binomial'] = {'params': {'n': 10, 'p': 0.6},
                            'dados': None,
                            'estatisticas': None}
```

```python
### Gerando os dados de cada distribuição

dados_poisson = np.random.poisson(1, 1000)
dados_binomial = np.random.binomial(10, 0.6, 1000)

distribuicoes['Poisson']['dados'] = dados_poisson
distribuicoes['Binomial']['dados'] = dados_binomial
```

```python
### Calculando as estatísticas básicas

mean_p = np.mean(distribuicoes['Poisson']['dados'])
median_p = np.median(distribuicoes['Poisson']['dados'])
var_p = np.var(distribuicoes['Poisson']['dados'])
desvio_p = np.std(distribuicoes['Poisson']['dados'])

mean_b = np.mean(distribuicoes['Binomial']['dados'])
median_b = np.median(distribuicoes['Binomial']['dados'])
var_b = np.var(distribuicoes['Binomial']['dados'])
desvio_b = np.std(distribuicoes['Binomial']['dados'])

distribuicoes['Poisson']['estatisticas'] = {'Média': float(mean_p), "Mediana": median_p, "Variância": var_p, "Desvio Padrão": desvio_p}
distribuicoes['Binomial']['estatisticas'] = {'Média': float(mean_b), "Mediana": median_b, "Variância": var_b, "Desvio Padrão": desvio_b}

dist = {'Normal': distribuicoes['Normal']['estatisticas'],
        'Exponencial': distribuicoes['Exponencial']['estatisticas'],
        'Uniforme': distribuicoes['Uniforme']['estatisticas'],
        'Poisson': distribuicoes['Poisson']['estatisticas'],
        'Binomial': distribuicoes['Binomial']['estatisticas']
        }
pd.DataFrame(dist)
```

# Passo 9 - Comparar distribuições graficamente

Para a criação de um único gráfico com os cinco boxplots, foi necessário acoplar todos os dados gerados de cada distribuição em uma única lista (matriz) para que a função nativa de boxplot pudesse trabalhar.

```python
matriz = []
matriz.append(distribuicoes['Normal']['dados'])
matriz.append(distribuicoes['Exponencial']['dados'])
matriz.append(distribuicoes['Uniforme']['dados'])
matriz.append(distribuicoes['Poisson']['dados'])
matriz.append(distribuicoes['Binomial']['dados'])

ax = sns.boxplot(matriz)
nomes_distribuicoes = ['Normal', 'Exponencial', 'Uniforme', 'Poisson', 'Binomial']
ax.set_xticklabels(nomes_distribuicoes)

plt.title('Boxplot das Distribuições', fontsize=14, pad=15)
plt.ylabel('Valores', fontsize=12, labelpad=10)
sns.despine()
plt.show()
```

# 10 - Comparar estatísticas calculadas com valores teóricos

A maior diferença observada ocorreu na distribuição Exponencial, apresentando uma discrepância significativa entre suas médias amostral e teórica, um comportamento que pode ser explicado pelo tamanho de n.

### Normal

```python
mean_teo_n = distribuicoes['Normal']['params']['loc']
var_teo_n = distribuicoes['Normal']['params']['scale']

dist_normal = {
    'Média': {"Amostra": round(mean_n, 3), "Teórica": round(mean_teo_n, 3)},
    'Variância': {"Amostra": round(var_n, 3), "Teórica": round(var_teo_n, 3)}
}

pd.DataFrame(dist_normal)
```

```python
nomes_barras = ['Média Amostral', 'Média Teórica']
valores_medias = [mean_n, mean_teo_n]

plt.bar(nomes_barras, valores_medias, width=0.5, color='lightblue')
sns.despine()
```

### Exponencial

```python
mean_teo_e = 1/(distribuicoes['Exponencial']['params']['scale'])
var_teo_e = 1/(distribuicoes['Exponencial']['params']['scale'])**2

dist_exp = {
    'Média': {"Amostra": round(mean_e, 3), "Teórica": round(mean_teo_e, 3)},
    'Variância': {"Amostra": round(var_e, 3), "Teórica": round(var_teo_e, 3)}
}

pd.DataFrame(dist_exp)
```

```python
nomes_barras = ['Média Amostral', 'Média Teórica']
valores_medias = [mean_e, mean_teo_e]

plt.bar(nomes_barras, valores_medias, width=0.5, color='olive')
sns.despine()
```

### Uniforme

```python
mean_teo_u = ((distribuicoes['Uniforme']['params']['low']) + (distribuicoes['Uniforme']['params']['high']))/2
var_teo_u = (((distribuicoes['Uniforme']['params']['high']) - (distribuicoes['Uniforme']['params']['low']))**2)/12

dist_uni = {
    'Média': {"Amostra": round(mean_u, 3), "Teórica": round(mean_teo_u, 3)},
    'Variância': {"Amostra": round(var_u, 3), "Teórica": round(var_teo_u, 3)}
}

pd.DataFrame(dist_uni)
```

```python
nomes_barras = ['Média Amostral', 'Média Teórica']
valores_medias = [mean_u, mean_teo_u]

plt.bar(nomes_barras, valores_medias, width=0.5, color='purple')
sns.despine()
```

### Poisson

```python
mean_teo_p = distribuicoes['Poisson']['params']['lambda']
var_teo_p = distribuicoes['Poisson']['params']['lambda']

dist_poi = {
    'Média': {"Amostra": round(mean_p, 3), "Teórica": round(mean_teo_p, 3)},
    'Variância': {"Amostra": round(var_p, 3), "Teórica": round(var_teo_p, 3)}
}

pd.DataFrame(dist_poi)
```

```python
nomes_barras = ['Média Amostral', 'Média Teórica']
valores_medias = [mean_p, mean_teo_p]

plt.bar(nomes_barras, valores_medias, width=0.5, color='mediumseagreen')
sns.despine()
```

### Binomial

```python
mean_teo_b = (distribuicoes['Binomial']['params']['n']) * (distribuicoes['Binomial']['params']['p'])
var_teo_b = ((distribuicoes['Binomial']['params']['n']) * (distribuicoes['Binomial']['params']['p'])) * (1-(distribuicoes['Binomial']['params']['p']))

dist_bin = {
    'Média': {"Amostra": round(mean_b, 3), "Teórica": round(mean_teo_b, 3)},
    'Variância': {"Amostra": round(var_b, 3), "Teórica": round(var_teo_b, 3)}
}

pd.DataFrame(dist_bin)
```

```python
nomes_barras = ['Média Amostral', 'Média Teórica']
valores_medias = [mean_b, mean_teo_b]

plt.bar(nomes_barras, valores_medias, width=0.5, color='gold')
sns.despine()
```
