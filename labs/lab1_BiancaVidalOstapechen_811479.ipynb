```python
def gerador(d):

  a = np.random.choice(2, d)

  seq2 = np.arange(len(a))

  teste= a * ((1/2)**(seq2+1))

  soma = sum(teste)

  return soma
```

```python
def DecomposiçãoBinaria(N):

  amostra = []

  for i in range(N):
    valor = gerador(N)
    amostra.append(float(valor))

  return amostra
```

```python
amostra_u = np.random.uniform(0,1, 1000)
print(amostra_u)
```

```python
amostra_DB = DecomposiçãoBinaria(1000)
print(amostra_DB)
```

```python
plt.hist(amostra_DB, alpha=0.5, label=' Decomposição Binária')
plt.hist(amostra_u, alpha=0.5, label="Distribuição Uniforme")

plt.title("Comparação entre Decomposição Binária e Distribuição Uniforme")
plt.legend(loc='upper right')
plt.show()
```

# Parte 2 do exercício 2

```python
### 2.1) Variável Bernoulli

p= 0.6

amostra_B = []

for i in amostra:
  if i <= p:
    amostra_B.append(1)
  else:
    amostra_B.append(0)

print(amostra_B)
```

```python
### 2.2)

def f_bernoulli(valores, p):
  amostra_final=[]
  for i in valores:
    if i <= p:
      amostra_final.append(1)
    else:
      amostra_final.append(0)
  return amostra_final

#p = 0.9
p = random.uniform(0,1)
print(p)
am = f_bernoulli(amostra, p)
print(am)
```

```python
### 2.3

mean = np.mean(am)
var = np.var(am)

mean_teorica = p
var_teorica = p*(1-p)

dist = {
    'Amostra': {"Média": mean, "Variância": var},
    'Teórica': {"Média": mean_teorica, "Variância": var_teorica}
}

pd.DataFrame(dist)
```

# 4/5 Distribuição Binomial

```python
### 4.1
def amostra_binomial(N, n, p):
    amostra_binomial=[]
    for i in range(N):
        amostra = np.random.uniform(0,1, n)
        soma = np.sum(f_bernoulli(amostra, p))
        amostra_binomial.append(soma)
    return np.array(amostra_binomial)

teste = amostra_binomial(10000, 1000, 0.1)
print(teste)
```

```python
### 4.2
n=1000
p=0.1

mean_t_b = n * p
var_t_b = n * p * (1 - p)

mean_amostra_b = np.mean(teste)
var_amostra_b = np.var(teste)

dist_binomial = {
    'Amostra': {"Média": round(mean_amostra_b, 3), "Variância": round(var_amostra_b, 3)},
    'Teórica': {"Média": mean_t_b, "Variância": var_t_b}
}

pd.DataFrame(dist_binomial)
```

# 5/5 Distribuição Multinomial

```python
### 5.1

def multinomial(amostra):
    amostra_amostrada = []
    for i in amostra:
        if i <= 0.1:
            amostra_amostrada.append(1)
        elif i <= 0.3:
            amostra_amostrada.append(2)
        elif i <= 0.6:
            amostra_amostrada.append(3)
        else:
            amostra_amostrada.append(4)
    return amostra_amostrada

teste2 = multinomial(amostra)
print(teste2)

teste2 = multinomial(amostra)

# Contando quantas vezes cada face apareceu
X1 = teste2.count(1)
X2 = teste2.count(2)
X3 = teste2.count(3)
X4 = teste2.count(4)

vetor_X = (X1, X2, X3, X4)
print(f"Lançamentos de cada face (X1, X2, X3, X4): {vetor_X}")
```

```python
### 5.2
import numpy as np

def simular_multinomial(N, n):
    vetores = []
    for i in range(N):
        amostra_uniforme = np.random.uniform(0, 1, n)
        faces_sor = multinomial(amostra_uniforme)

        X1 = faces_sor.count(1)
        X2 = faces_sor.count(2)
        X3 = faces_sor.count(3)
        X4 = faces_sor.count(4)
        
        vetor_X = [X1, X2, X3, X4]
        vetores.append(vetor_X)
        
    return np.array(vetores)

matriz = simular_multinomial(1000, 100)
```

```python
### 5.3

n = 100
prob = np.array([0.10, 0.20, 0.30, 0.40])

means_t = (n*prob)
print("Médias Teóricas:", means_t) 

means_amostrais = np.mean(matriz, axis=0)

medias_amostrais = np.round(means_amostrais, 2)
print(f"Médias Amostrais:", medias_amostrais)
```

```python
### 5.4

segunda_amostra_mult = np.random.multinomial(n, prob, 1000)
mean_segunda_amostra_mult = np.mean(segunda_amostra_mult, axis=0)

print("Médias Teóricas:", means_t) 
print("Médias Amostrais:", medias_amostrais)
print("Médias da Função Pronta:", (mean_segunda_amostra_mult))
```

```python
### 5.5
somas_manuais = np.sum(matriz, axis=1)
somas_função = np.sum(segunda_amostra_mult, axis=1)

verificando= np.all(somas_manuais == n)
verificando_função= np.all(somas_função == n)

print("Amostra:", verificando)
print("Função Pronta:", verificando_função)
```

```python
### 5.6
var_teoricas = n * prob * (1 - prob)
var_amostrais = np.var(matriz, axis=0, ddof=1)

dist_mult= {
    'Face': [1, 2, 3, 4],
    'Variância Teórica': np.round(var_teoricas, 4),
    'Variância Amostral': np.round(var_amostrais, 4)
}

pd.DataFrame(dist_mult)
```

```python
### 5.7

p1 = 0.10
p2 = 0.20


cov_teorica = -n * p1 * p2
face1_amostral = matriz[:, 0]
face2_amostral = matriz[:, 1]

matriz_cov = np.cov(face1_amostral, face2_amostral, ddof=1)
cov_amostral = matriz_cov[0, 1]

print(f"Teórica:  {cov_teorica:.4f}")
print(f"Amostral: {cov_amostral:.4f}")
```

O aumento na contagem de uma categoria força a diminuição na contagem de outra, gerando uma relação inversamente proporcional.
